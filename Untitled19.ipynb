```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('twitter_training.csv',header = None)
```

```python
print(dataset)
```

```python
import nltk
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
for column in dataset.select_dtypes(include=['object']).columns:
  dataset[column] = dataset[column].apply(lambda x : nltk.word_tokenize(str(x)))

  dataset.to_csv("tokenized_file.csv",index = False)
```

```python
ds = pd.read_csv('tokenized_file.csv',header=None )
ds.columns = ['tokenize', 'other_column1', 'other_column2','text']
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import string
nltk.download('wordnet')
nltk.download('stopwords')
```

```python
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
```

```python
def preprocess(text):
  tokens = [word for word in text.lower().split()]
  tokens = [word for word in tokens if word.isalpha()]
  tokens = [word for word in tokens if word not in stop_words]
  tokens = [lemmatizer.lemmatize(word) for word in tokens]
  return tokens
ds['token'] = ds['text'].apply(preprocess)
print(ds['token'].head())
```

```python
ds.to_csv("final.csv",index=False)
```
