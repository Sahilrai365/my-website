```python
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
from nltk import pos_tag

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

da = pd.read_csv('IMDB_Review.csv',on_bad_lines='skip',encoding='utf-8')

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

def preprocess_text(text):
    words = word_tokenize(text.lower())
    words = [word for word in words if word.isalnum()]
    words = [word for word in words if word not in stop_words]
    pos_tags = pos_tag(words)
    words = [lemmatizer.lemmatize(word, get_wordnet_pos(tag)) for word, tag in pos_tags]
    return " ".join(words)


da['review'] = da['review'].astype(str).apply(preprocess_text)


da.to_csv('lemmatized_IMDB_Review.csv', index=False)

print("Lemmatization with POS tagging applied successfully!")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer(max_features=5000)
X = vectorizer.fit_transform(da['review'])
```

```python
da['sentiment'] = da['sentiment'].map({'positive': 1, 'negative': 0})
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
X_train, X_test, y_train, y_test = train_test_split(X, da['sentiment'], test_size=0.2, random_state=42)
rf_model = RandomForestClassifier(n_estimators=100,class_weight="balanced", random_state=42)
rf_model.fit(X_train, y_train)
y_pred = rf_model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy}")
```

```python
# Function to predict sentiment for a new review
def predict_sentiment(review):
    review_tfidf = vectorizer.transform([review])  # Transform text using trained TF-IDF
    prediction = rf_model.predict(review_tfidf)  # Predict using trained model
    return "Positive" if prediction[0] == 1 else "Negative"

# Example Predictions
print(predict_sentiment("Worst movie I ever saw"))
print(predict_sentiment("A masterpiece! The cinematography was beautiful."))
```
